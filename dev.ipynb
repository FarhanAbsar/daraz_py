```python
import json

order_info = {}

order_ids = df['order_no'].unique()

for order_id in order_ids:
    # order = discount.get_order(order_id)
    request = iop.IopRequest('/order/get', 'GET')

    request.add_api_param('access_token', access_token)
    request.add_api_param('order_id', order_id)

    response = client.execute(request, access_token)

    order_data_json = response.body
    order = json.loads(order_data_json)["data"]
    print(order)
    order_info[
        order_id
    ] = {
        "created_at": order["created_at"],
        "retail_price" : order["price"] if "price" in order else None,
        "voucher" : order["voucher"] if "voucher" in order else 0.0,
    } 

# order_info
```

```python
# 1️⃣ Clean amount column FIRST
df['amount'] = (
    pd.to_numeric(df['amount'].astype(str).str.replace(',',''), errors='coerce')  # '' or junk → NaN
      .fillna(0)                                  # NaN → 0
)

# define bill amount (only product price rows)
df['Bill Amt'] = np.where(
    df['fee_type'] == "13",
    df['amount'],
    0
)

# define discount base (all others)
df['Discount Base'] = np.where(
    df['fee_type'] != "13",
    df['amount'],
    0
)
```

```python
df[df['order_no']=='683734719067225']
```

```python
# group by order_no
new_df = (
    # df.groupby('transaction_number', as_index=False)
    df.groupby('order_no', as_index=False)
      .agg({
          'order_no': 'first',
          'Bill Amt': 'sum',
          'Discount Base': 'sum',
        #   'transaction_date': 'last'
      })
)

# rename columns
new_df = new_df.rename(columns={
    'order_no': 'Daraz Bill No.',
    'transaction_date': 'Date'
})

# discount amount = total other amounts × (-1)
new_df['Discount Amt'] = new_df['Discount Base'] * -1

# new_df['Discount Amt'] = (
#     pd.to_numeric(
#         new_df["Discount Amt"], errors='coerce'
#     )
#     .fillna(0)
#     .astype(int)
# )
# new_df['Discount Amt'].fillna(0, inplace=True)
# new_df['Discount Amt'].where('') = 0
# new_df['Discount Amt'] = new_df['Discount Amt'].astype(int)

# received amt

# empty columns
new_df['Payment Amt'] = ''
new_df['Remarks'] = ''
new_df['Bill No.'] = ''
new_df['Order Created At'] = new_df["Daraz Bill No."].apply(lambda x : order_info[x]["created_at"])

new_df["Discount Amt"] = new_df.apply(
    lambda x : 
    (
        x["Discount Amt"]
        + float(
            str(
                order_info[x["Daraz Bill No."]]["retail_price"]
                # order_info[x["Daraz Bill No."]]["voucher"]
            ).replace(',','')
        )
        - x["Bill Amt"]
    ) if (
        x["Bill Amt"] != 0.0
    )
    else x["Discount Amt"]
    , 
    axis=1
)

new_df["Discount Amt"] = new_df["Discount Amt"].round(2)

new_df["Bill Amt"] = new_df.apply(
    lambda x : 
    float(order_info[x["Daraz Bill No."]]["retail_price"]) 
    if (
        # (float(x["Bill Amt"]) != float(order_info[x["Daraz Bill No."]]["retail_price"]))
        (float(order_info[x["Daraz Bill No."]]["voucher"]) != 0.0 )
        and (x["Bill Amt"] != 0.0)
    )
    else x["Bill Amt"], 
    axis=1
)

new_df['Received Amt'] = new_df['Bill Amt'] - new_df['Discount Amt']

new_df['Received Amt'] = new_df['Received Amt'].round(2)

print(
    type(new_df["Discount Amt"][0]),
    type(new_df["Bill Amt"][0]),
    type(new_df["Received Amt"][0]),
)
# drop helper column
new_df = new_df.drop(columns=['Discount Base'])

new_df = new_df[[
    'Order Created At', 'Date', 'Payment Amt', 'Bill No.', 
    'Daraz Bill No.', 'Bill Amt', 'Received Amt', 'Discount Amt', 'Remarks'
]]
```
